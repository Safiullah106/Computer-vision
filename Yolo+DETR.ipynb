```python
import os
from ultralytics import YOLO

# 1. YOLOv8 ka Medium model load karein (accurate aur fast)
model = YOLO('yolov8m.pt')

# 2. Aapki video ka path
input_video = "/content/drive/MyDrive/cars.mp4"

# 3. Google Drive mein save karne ka folder path
output_dir = "/content/drive/MyDrive/AI_Outputs"
os.makedirs(output_dir, exist_ok=True)

print("AI aapki video 'cars.mp4' ko process kar raha hai... Please wait...")

# 4. Object tracking aur detection run karein
model.track(
    source=input_video,
    conf=0.3,             # 30% confidence threshold (sirf confirm objects pakray ga)
    persist=True,          # Vehicles ko track karne ke liye (ID assign karega)
    save=True,             # Video save karne ke liye
    project=output_dir,    # Drive ka folder jahan save karna hai
    name="cars_tracked"    # Sub-folder ka naam
)

print("\nProcessing Complete! 🎉")
print(f"Aapki processed video aapki Google Drive mein is folder mein save ho chuki hai:\n{output_dir}/cars_tracked")
```

```python
import cv2
import os
from ultralytics import RTDETR

# 1. RT-DETR Large model load karein
model = RTDETR('rtdetr-l.pt')

# 2. Paths set karein
input_video_path = "/content/drive/MyDrive/trafic.mp4"
output_video_path = "/content/drive/MyDrive/AI_Outputs/cars_rtdetr_fixed.mp4"

# Ensure output directory exists
os.makedirs(os.path.dirname(output_video_path), exist_ok=True)

# 3. OpenCV Video capture aur setup
cap = cv2.VideoCapture(input_video_path)
assert cap.isOpened(), "Video file open nahi ho rahi!"

w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))

# Video Writer setup (Saving frames back to mp4)
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
video_writer = cv2.VideoWriter(output_video_path, fourcc, fps, (w, h))

print("RT-DETR (Transformer) memory-efficient loop mein chal raha hai...")

# 4. Stream=True use kar ke frame-by-frame predict karna (No RAM bottleneck)
# stream=True lagane se memory save hoti hai aur RAM error nahi aata
results = model.predict(source=input_video_path, conf=0.3, stream=True)

for r in results:
    # Bounding boxes aur labels draw kiye hue frame ko nikalna
    annotated_frame = r.plot()

    # Processed frame ko video file mein save karna
    video_writer.write(annotated_frame)

# 5. Resources release kar dein safely
cap.release()
video_writer.release()
cv2.destroyAllWindows()

print(f"\n🎉 Done! Output video perfectly save ho chuki hai:\n{output_video_path}")
```
